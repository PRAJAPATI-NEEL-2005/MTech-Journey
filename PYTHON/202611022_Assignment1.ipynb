1. Accept an integer and determine whether it is:

Armstrong Number
Palindrome Number
Perfect Number
Display every property satisfied.

```python
num = int(input("Enter an integer: "))
armstrong = False
palindrome = False
perfect = False
temp = num
size = len(str(num))
sum = 0

while temp > 0:
    rem = temp % 10
    sum += rem ** size
    temp //= 10

if sum == num:
    armstrong = True

temp = num
reverse = 0
while temp > 0:
    rem = temp%10
    reverse = reverse * 10 + rem
    temp //= 10
if reverse == num:
    palindrome = True

divisorsum = 0
for i in range(1, num // 2 + 1):
    if num % i == 0:
        divisorsum += i
if divisorsum == num:
    perfect = True


if armstrong:
    print("- Armstrong Number")
else:
    print("- Not Armstrong Number")

if palindrome:
    print("- Palindrome Number")
else:
    print("- Not Palindrome Number")

if perfect:
    print("- Perfect Number")
else:
    print("- Not Perfect Number")

if armstrong and palindrome and perfect:
    print("- every property satisfied")

```

2. Implement the Euclidean Algorithm to compute the GCD of two integers. Then compute
the LCM using the computed GCD.

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
num1 = a
num2 = b
while b != 0:
    a, b = b, a % b
gcd = a
lcm = (num1 * num2) // gcd
print("GCD =", gcd)
print("LCM =", lcm)
```

5. Given L = [10, 20, 30, [40, 50]], write code to check whether 40 is a direct member of L
(should print False since it's nested) and whether it exists at all if you flatten one level.

```python
L=[10, 20, 30, [40, 50]]
print(40 in L)
flat=[]
for item in L:
    if type(item)==list:
        for x in item:
            flat.append(x)
    else:
        flat.append(item)
print(40 in flat)
```

8. Write a program that takes a list of numbers and a target value, and uses in/not in to
classify which target values exist and which don't — without using if x in list: print(...)
directly; instead, build a small membership-report function.

```python
def membershipreport(numbers, target):
    if target in numbers:
        return f"{target} exists in the list."
    else:
        return f"{target} does not exist in the list."

numbers = list(map(int, input("Enter list elements separated by space: ").split()))

n = int(input("How many target values? "))

for i in range(n):
    target = int(input("Enter target value: "))
    print(membershipreport(numbers, target))
```

```python
a = -1
b = 5

if a > 0:
    if b > 0:
        print("both positive")
else:
    print("a not positive")

"""
Explanation:
Here, a = -1 and b = 5.
The condition (a > 0) is False.
Therefore, Python skips the entire nested if block.
The else is aligned with the outer 'if a > 0', so it belongs to that if.
Hence, "a not positive" is printed.

Python has no ambiguity because it uses indentation to determine
which if an else belongs to. An else always matches the nearest
preceding if at the same indentation level.
"""
```

12. WAP to find the sum of a number until it becomes a single digit. (example: 9875 → 29→
11→ 2)

```python
num = int(input("Enter a number: "))
while num >= 10:
    digit_sum = 0
    while num > 0:
        digit_sum += num % 10
        num //= 10
    print("Sum =", digit_sum)
    num = digit_sum
print("Single digit =", num)
```

13. WAP to search for a target element in a list. DO NOT use a separate flag variable -rely
purely on ‘while-else’ semantics (loop completed without break)

```python
numbers = list(map(int, input("Enter list elements: ").split()))
target = int(input("Enter target element: "))
i = 0
while i < len(numbers):
    if numbers[i] == target:
        print("Element found at index", i)
        break
    i += 1
else:
    print("Element not found")
```

14. WAP using ‘for-else’ to check if a list contains any duplicate elements. Print “No
duplicates found” if the loop runs to completion.

```python
numbers = list(map(int, input("Enter list elements: ").split()))
for i in range(len(numbers)):
    for j in range(i + 1, len(numbers)):
        if numbers[i] == numbers[j]:
            print("Duplicate found:", numbers[i])
            break
    else:
        continue
    break
else:
    print("No duplicates found")
```

16. Generate a secret number inside the program. Keep asking the user until the correct
number is entered. Display the number of attempts.

```python
secret = 25     
attempts = 0

while True:
    guess = int(input("Enter your guess: "))
    attempts += 1

    if guess == secret:
        print("Correct! You guessed the secret number.")
        print("Number of attempts:", attempts)
        break

    elif guess > secret:
        print("Your guess is greater than the secret number. Try again!")

    else:
        print("Your guess is smaller than the secret number. Try again!")
```

17. WAP to print Floyd’s Triangle numbers up to n rows.

```python
n = int(input("Enter number of rows: "))
num = 1
for i in range(1, n + 1):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```

20. WAP for:
* * * * *
*       *
*       *
*       *
* * * * *

```python
rows = int(input("Enter the number of rows: "))
cols = int(input("Enter the number of columns: "))

for i in range(rows):
    for j in range(cols):

        if i == 0 or i == rows - 1 or j == 0 or j == cols - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print()
```
