### 2. Hotel booking calculator - positional, keyword n default
A hotel calculates the final bill based on:\
● Number of rooms\
● Number of nights\
● Room price\
● Breakfast charges\
● Extra-bed charges\
● Discount\
Some charges are optional, and most bookings use standard values.
The program should allow a booking to be created using different combinations of
supplied information.

```python
roomchargepernight=2500
breakfast=300
extrabed=500
discount=20

def calculatebill(no_rooms,no_nightstay,no_extrabed,breakfastinclude):
  roomcost=roomchargepernight*no_rooms*no_nightstay
  if breakfastinclude:
    breakfastcost=no_nightstay*breakfast
  else:
    breakfastcost=0
  extrabedcost=no_extrabed*extrabed
  totalcost=((roomcost+breakfastcost+extrabedcost)-(roomcost+breakfastcost+extrabedcost)*discount/100)
  return totalcost

no_rooms=int(input("how many rooms do you want:"))
no_nightstay=int(input("for how many nights you want to book?:"))
no_extrabed=int(input("give no of extrabeds if you want:"))
choice = input("Do you want to include brakfast? (yes/no): ").lower().strip()
if choice in ['yes', 'y']:
  breakfastinclude= True
elif choice in ['no', 'n']:
  breakfastinclude = False

print("your total bill cost :",calculatebill(no_rooms,no_nightstay,no_extrabed,breakfastinclude))

```

### 3. Employee Salary & Benefits System - *args, **kwargs

A company wants a reusable function to generate an employee's salary report.
Every employee has:\
● Employee name\
● Basic salary\
● Any number of allowances, where only the amounts matter\
● Any number of deductions, where each deduction has a meaningful name such
as PF, Tax, Insurance, Loan

Calculate and display:\
● Total allowances\
● Gross salary\
● Total deductions\
● Net salary\
● Complete deduction details

```python
def salaryreport(ename,basicsalary,*allowances,**deductions):
  totalallowance=sum(allowances)
  grosssalary=basicsalary+totalallowance
  totaldeduction=0
  for dedname,amount in deductions.items():
    totaldeduction+=amount
  netsalary=grosssalary-totaldeduction

  print("\n SALARY REPORT FOR ",ename)
  print("Total allowances :",totalallowance)
  print("Gross salary:",grosssalary)
  print("Total deductions:",totaldeduction)
  print("Net salary:",netsalary)
  print("deduction  details are shown below")
  if deductions:
    for dedname,amount in deductions.items():
      print(dedname,"=",amount)
  else:
    print("there is not any deductions")


ename=input("enter employee name:")
basic = float(input("Enter basic salary: "))
allowances = []
n=int(input("enter no of allowance you wanted to add"))
for i in range(0,n):
  allow=int(input(f"allowance {i+1}"))
  allowances.append(allow)
deductions={}
k=int(input("enter no of deductions you wanted to add:"))
print("first enter the deductionnmae and after that enter the value to be deducted")
for i in range(0,k):
  dedname=input(f"enter deduction {i+1} name:")
  dedvalue=int(input(f"enter value  of {dedname}:"))
  deductions[dedname]=dedvalue

salaryreport(ename,basic,*allowances,**deductions)
```

### 4. Number Group Analyzer - Nested fun, Recursive fun, lambda, *args
A mathematics application accepts a list of positive integers from the user. The
application needs to analyze the numbers and produce a summary.
For the giveDisplay the numbers in ascending order.\
● Display only the numbers that are even.\
● Calculate the factorial of each number.\
● Display the sum of all factorial values.\
● Display the largest factorial value.\
● Display the final analysis in a well-organized format.\
● n list, the program should:

```python


def analyzenumbers(*args):
    def factorial(n):
        if n == 0 or n == 1:
            return 1
        return n * factorial(n - 1)
    ascending=sorted(args, key=lambda x: x)
    evennumbers=list(filter(lambda x: x % 2 == 0, args))
    factorials=list(map(lambda x: factorial(x), args))
    total=sum(factorials)
    largest =max(factorials)
    print("Original Numbers:",list(args))
    print("Ascending Order :",ascending)
    print("Even Numbers    :",evennumbers)
    print("\nFactorials:")
    for i in range(len(args)):
        print(args[i], "! =", factorials[i])
    print("\nSum of Factorials    :", total)
    print("Largest Factorial   :", largest)




numbers = list(map(int, input("Enter positive integers separated by space: ").split()))


analyzenumbers(*numbers)
```

### 6. Employee Bonus & Performance Tracker — Scope
A company wants to maintain an employee's performance information and calculate a bonus based on performance.\

The program must:\
● Store the company bonus percentage so that it is available to the
salary-calculation function.\
● Calculate the employee's bonus based on the performance score.\
● Display the employee's basic salary, bonus percentage, bonus amount, and final
salary.\

reflect the updated value in subsequent calculations.\
● Create an internal/helper operation for calculating the bonus and demonstrate
how it accesses values defined outside its own local scope.\
● Ensure that changing a local variable with the same name as the company bonus
percentage does not unintentionally change the company-wide value.

```python
bonusper = 10
def salary(name, basicsal, score):
    def bonus():
        if score >= 90:
            b = basicsal * bonusper / 100
        elif score >= 75:
            b = basicsal * (bonusper - 2) / 100
        elif score >= 50:
            b = basicsal * (bonusper - 5) / 100
        else:
            b = 0
        return b
    b = bonus()
    finalsal = basicsal + b
    print("\nEmployee Name:", name)
    print("Basic Salary:", basicsal)
    print("Performance Score:", score)
    print("Bonus Percentage:", bonusper, "%")
    print("Bonus Amount:", b)
    print("Final Salary:", finalsal)

def changebonus(x):
    global bonusper
    bonusper = x
    print("\nBonus percentage changed to", bonusper, "%")

def testlocal():
    bonusper = 50
    print("\nInside function bonus percentage:", bonusper)


name = input("Enter employee name: ")
basicsal = float(input("Enter basic salary: "))
score = float(input("Enter performance score: "))
salary(name, basicsal, score)
testlocal()
print("Company bonus percentage:", bonusper)
x = float(input("\nEnter new bonus percentage: "))
changebonus(x)
print("\nAfter changing bonus percentage:")
salary(name, basicsal, score)
```
