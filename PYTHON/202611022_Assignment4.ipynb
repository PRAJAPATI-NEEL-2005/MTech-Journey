##1. Create a smart-home application through Python that manages devices such as lights, fans, and air conditioners.
Create a class ‘SmartDevice’ with information about the device name, room, power
consumption, and current status. [Constructor, Instance Attributes]
The system should:\
● Prevent invalid power-consumption values.\
● Allow the device to be switched ON or OFF.\
● Calculate the estimated power usage based on the number of operating hours.\
● Display a readable description whenever the object is printed. [__str__()]\
● Provide a detailed representation suitable for inspecting objects inside a list.\
[__repr__()]\
The application should also maintain the total number of devices registered in the
smart-home system. [Class Attribute]
Create at least four devices and demonstrate the effect of modifying their individual
status and the shared device count.

```python
class Smartdevices:
  totaldevice=0
  def __init__(self,name,room,power_consumption):
    self.name=name
    self.room=room
    if power_consumption<0:
      raise ValueError("Power consumption cannot be negative.")
    self.power_consumption=power_consumption
    self.status = False
    Smartdevices.totaldevice += 1

  def turnon(self):
    self.status=True

  def turnoff(self):
    self.status=False

  def powerusage(self,hours):
    if hours<0:
      raise ValueError("hours can not be negative")
    usage=self.power_consumption*hours
    return usage

  def __str__(self):
    state = "ON" if self.status else "OFF"
    return f"{self.name} in {self.room} | Status: {state} | Power: {self.power_consumption}W"

  def __repr__(self):
    return f"SmartDevice(name='{self.name}', room='{self.room}', power_consumption={self.power_consumption}, status={self.status})"


device1 = Smartdevices("Main Light", "Living Room", 15)
device2 = Smartdevices("Ceiling Fan", "Bedroom", 75)
device3 = Smartdevices("Air Conditioner", "Bedroom", 1500)
device4 = Smartdevices("Desk Lamp", "Office", 10)

print(f"\nTotal devices registered in the system: {Smartdevices.totaldevice}")

device2.turnon()
device3.turnon()
device1.turnon()
device1.turnoff()
device3.turnoff()


hours_used = 8
acusage=device3.powerusage(hours_used)
print(f"\nEstimated power usage for {hours_used} hours: {acusage}W")

print("device 2 condition:")
print(device2)

print("device 4 conition")
print(device4)

home_network = [device1, device2, device3, device4]
print(home_network)
```

## 2. Python Programme for software application temporarily creates objects that represent system resources, such as database connections or application sessions.
Create a class Resource that stores: Resource name, Resource type, Resource ID.
When an object is created, display a message indicating that the resource has
become active.\
When an object is destroyed, display a message indicating that the resource has
been released. [__del__()]\
The class should also:\
● Reject an empty resource ID. [assert]\
● Provide a meaningful display when printed. [__str__()]\
● Provide a detailed representation for debugging and collection display.\
[__repr__()]\
Create several resources, explicitly remove selected objects using del, and observe
the resulting lifecycle messages.

```python
class Resources:
  def __init__(self,resname,restype,resid):
    assert resid != "", "Resource ID cannot be empty!"
    self.resname=resname
    self.restype=restype
    self.resid=resid
    print(f"{self.resname} has become active")

  def __del__(self):
    print(f"{self.resname} has been released")

  def __str__(self):
    return f"{self.resname} | {self.restype} | {self.resid}"

  def __repr__(self):
    return f"Resources(resname='{self.resname}', restype='{self.restype}', resid='{self.resid}'"


db_conn = Resources("Primary_DB", "Database Connection", "DB-001")
session = Resources("User_Admin", "App Session", "SESS-409")
cache = Resources("Redis_Node", "Cache Allocation", "CACHE-992")

print("displayed objects using __str__() method:")
print(db_conn)
print(session)
print(cache)

print("displayed objects using __repr__() method:")
resource_list = [db_conn, session, cache]
print(resource_list)


```

##3. WAP for - A university system may create different Python objects representing the same student. The system needs a meaningful way to determine whether two student
objects refer to the same student.\
Create a class StudentProfile containing: Student ID, Name, Program, Semester.
Two student objects should be considered equal when they have the same student
ID, even if their other information is different. [__eq__() – learn the dunder method
and use]\
Also provide:\
● A readable output when an individual object is printed. [__str__()]\
● A detailed representation when objects are inspected inside a collection.
[__repr__()]\
Create at least four student objects and compare different pairs using ==.

```python
class StudentProfile:
  def __init__(self,student_id,name,program,semester):
    self.student_id=student_id
    self.name=name
    self.program=program
    self.semester=semester

  def __eq__(self,other):
    return self.student_id==other.student_id

  def __str__(self):
    return f"{self.name} [ID: {self.student_id}] - {self.program} (Semester {self.semester})"

  def __repr__(self):
    return f"StudentProfile(student_id='{self.student_id}', name='{self.name}', program='{self.program}', semester={self.semester})"



student1=StudentProfile("1","neel","cs",1)
student2=StudentProfile("2","krushna","ml",1)
student3=StudentProfile("3","om","ss",1)
student4=StudentProfile("1","viraj","ml",1)
student5=StudentProfile("4","rushabh","ss",1)

print("displayed objects using __str__() method:")
print(student1)
print(student2)
print(student3)
print(student4)
print(student5)


print("displayed objects using __repr__() method:")
print([student1,student4])


print(f"student1 is same as student4: {student1==student4}")
print(f"student1 is same as student5: {student1==student5}")
print(f"student4 is same as student5: {student4==student5}")

```

##4. Develop E-Commerce Product Catalogue by python prog. Lang. for an e-commerce platform maintains a catalogue of products sold by different sellers.
Create a class Product containing: Product ID, Product name, Category, Price, Stock
quantity\
The application should:\
● Reject invalid prices and stock quantities.\
● Allow stock to be increased or decreased.\
● Calculate the current inventory value of a product.\
● Maintain a common platform name for all products.\
● Keep track of the number of product objects registered through a class-level
mechanism.\
● Provide meaningful output when individual products are printed.\
● Provide detailed representations when a collection of products is displayed.\
Task: Create at least five products, modify stock for multiple products, display their
inventory values, and demonstrate the difference between class-level and
object-level information.

```python
class Product:
  platform_name = "Global E-Commerce Hub"
  total_products = 0

  def __init__(self,id,name,category,price,quantity):
    self.prodid=id
    self.prodname=name
    self.category=category
    if price<0:
      raise ValueError("Price cannot be negative")
    self.price=price
    if quantity<0:
      raise ValueError("Stock quantity cannot be negative")
    self.stockquantity=quantity
    Product.total_products += 1

  def stockincrease(self,quantity):
    if quantity<=0:
      raise ValueError("Stock quantity cannot be negative or zero")
    self.stockquantity+=quantity

  def stockdecrease(self,quantity):
    if quantity<=0:
      raise ValueError("Stock quantity cannot be negative or zero")
    if self.stockquantity<quantity:
      raise ValueError("Insufficient stock")
    self.stockquantity-=quantity

  def inventoryvalue(self):
    return self.price*self.stockquantity

  def __str__(self):
    return f"{self.prodname} [ID: {self.prodid}] - {self.category} | Price: ${self.price} | Stock: {self.stockquantity}"

  def __repr__(self):
    return f"Product(prodid='{self.prodid}', prodname='{self.prodname}', category='{self.category}', price={self.price}, stockquantity={self.stockquantity})"


product1=Product("1","mobile","electronics",10000,10)
product2=Product("2","laptop","electronics",50000,5)
product3=Product("3","charger","electronics",1000,10)
product4=Product("4","mouse","electronics",600,15)
product5=Product("5","lamp","home decore",1500,25)


print("inventory value for all products before increase or decrease are as below:")
for p in [product1,product2,product3,product4,product5]:
  print(f"{p.prodname}:",p.inventoryvalue())
product1.stockdecrease(5)
product2.stockincrease(10)
product3.stockdecrease(1)
product4.stockdecrease(2)
product5.stockdecrease(5)

print("inventory value for all products after increase or decrease of stock are as below:")
for p in [product1,product2,product3,product4,product5]:
  print(f"{p.prodname}:",p.inventoryvalue())

print(f"Total products registered: {Product.total_products}")

print("displayed objects using __str__() method:")
print(product1)
print(product2)
print(product3)
print(product4)
print(product5)

print("displayed product catalog using __repr__() method:")
catalog = [product1, product2, product5]
print(catalog)



print("object level info")
print(f"product1: {product1.__dict__}")
print(f"product2: {product2.__dict__}")

print("class level info")
print(f"Product: {Product.__dict__}")
```

```python
class Digitalresources:

  university="DAIICT"

  def __init__(self,resid,name,owner,size):
    if not resid or not name or not owner or not size:
      raise ValueError("name resouce id and owner cannot be empty")
    if size<=0:
      raise ValueError("size cannot be zero or negative")
    self.resid=resid
    self.name=name
    self.owner=owner
    self.size=size

  def resupdate(self,newname=None,newsize=None):
    if newname is not None:
      self.name=newname
    if newsize is not None:
      if newsize<=0:
        raise ValueError("size cannot be zero or negative")
      self.size=newsize

  def __str__(self):
    return f"{self.name} [ID: {self.resid}] - {self.owner} | Size: {self.size}MB"

  def __repr__(self):
    return f"Digitalresources(resid='{self.resid}', name='{self.name}', owner='{self.owner}', size={self.size})"

  def __eq__(self,other):
    return self.resid==other.resid

  def __del__(self):
    print(f"{self.name} has been released")


resource1=Digitalresources("1","python","neel",10)
resource2=Digitalresources("2","java","krushna",15)
resource3=Digitalresources("3","c++","om",20)
resource4=Digitalresources("4","c","viraj",25)
resource5=Digitalresources("5","cpp","rushabh",30)

resource1.resupdate(newsize=10)
resource5.resupdate(newname="c#",newsize=35)


print("displayed objects using __str__() method:")
print(resource1)
print(resource2)
print(resource3)
print(resource4)
print(resource5)
print()

print("displayed objects using __repr__() method:")
resource_list = [resource1, resource2, resource3, resource4, resource5]
print(resource_list)
print()


print(f"resource1 is same as resource2: {resource1==resource2}")
print(f"resource1 is same as resource5: {resource1==resource5}")
print()

print(Digitalresources.university)
Digitalresources.university="DAU"
print(Digitalresources.university)
print()


```
