### TASK 5: Hospital Patient Management System

```python
# A simple Patient class to hold individual details
class Patient:
    def __init__(self, pid, name, age, bgroup, econtact):
        self.pid = pid
        self.name = name
        self.age = age
        self.bgroup = bgroup
        self.econtact = econtact

        # Lists to store multiple items
        self.disease_list = []
        self.phone_list = []

    def add_disease(self, dname):
        # Only add disease if it is not already in the list
        if dname not in self.disease_list:
            self.disease_list.append(dname)

    def remove_disease(self, dname):
        if dname in self.disease_list:
            self.disease_list.remove(dname)
            print("Disease removed successfully.")
        else:
            print("Disease not found in patient record.")

    def add_phone(self, phone_no):
        self.phone_list.append(phone_no)

    def show_details(self):
        print("-----------------------------------")
        print("Patient ID:", self.pid)
        print("Name:", self.name)
        print("Age:", self.age)
        print("Blood Group:", self.bgroup)
        print("Emergency Contact:", self.econtact)
        print("Contact Numbers:", self.phone_list)
        print("Diseases:", self.disease_list)
        print("-----------------------------------")


# Hospital management class
class HospitalSystem:
    def __init__(self):
        # Dictionary to store all patients: key = pid, value = Patient object
        self.all_patients = {}

        # Tuple for valid blood groups (Fixed/Immutable)
        self.valid_blood_groups = ("A+", "A-", "B+", "B-", "AB+", "AB-", "O+", "O-")

    # 1. Add new patient
    def add_new_patient(self, pid, name, age, bgroup, econtact):
        if pid in self.all_patients:
            print("Error: Patient ID already exists!")
            return

        # Checking if blood group is valid
        if bgroup not in self.valid_blood_groups:
            print("Error: Invalid blood group!")
            return

        # Creating new patient object
        new_p = Patient(pid, name, age, bgroup, econtact)
        self.all_patients[pid] = new_p
        print("Patient added successfully!")

    # 2. Search for a patient
    def search_patient(self, pid):
        if pid in self.all_patients:
            return self.all_patients[pid]
        else:
            print("Patient not found!")
            return None

    # 3. Update patient information
    def update_patient(self, pid, new_name, new_age, new_econtact):
        p = self.search_patient(pid)
        if p != None:
            if new_name != "":
                p.name = new_name
            if new_age != 0:
                p.age = new_age
            if new_econtact != "":
                p.econtact = new_econtact
            print("Patient information updated!")

    # 6. Display all different diseases in hospital
    def display_all_diseases(self):
        hospital_diseases = []
        for pid in self.all_patients:
            p = self.all_patients[pid]
            for d in p.disease_list:
                if d not in hospital_diseases:
                    hospital_diseases.append(d)

        print("All Diseases Recorded in Hospital:")
        print(hospital_diseases)

    # 7. Find patients with a particular disease
    def find_patients_by_disease(self, dname):
        print("Patients suffering from", dname, ":")
        found = False
        for pid in self.all_patients:
            p = self.all_patients[pid]
            if dname in p.disease_list:
                print("-", p.name, "(ID:", p.pid, ")")
                found = True
        if found == False:
            print("No patients found with this disease.")

    # 8. Find patients with a particular blood group
    def find_patients_by_blood_group(self, bgroup):
        print("Patients with Blood Group", bgroup, ":")
        found = False
        for pid in self.all_patients:
            p = self.all_patients[pid]
            if p.bgroup == bgroup:
                print("-", p.name, "(ID:", p.pid, ")")
                found = True
        if found == False:
            print("No patients found with this blood group.")

    # 11. Display complete info of all patients
    def display_all_patients(self):
        if len(self.all_patients) == 0:
            print("No patient records available.")
        else:
            for pid in self.all_patients:
                p = self.all_patients[pid]
                p.show_details()

    # 12. Delete a patient
    def delete_patient(self, pid):
        if pid in self.all_patients:
            del self.all_patients[pid]
            print("Patient deleted successfully!")
        else:
            print("Patient ID not found!")

    # 13. Find patient with highest number of diseases
    def patient_with_most_diseases(self):
        if len(self.all_patients) == 0:
            print("No patients in hospital.")
            return

        max_count = -1
        top_patient = None

        for pid in self.all_patients:
            p = self.all_patients[pid]
            num_diseases = len(p.disease_list)
            if num_diseases > max_count:
                max_count = num_diseases
                top_patient = p

        if top_patient != None:
            print("Patient with most diseases is", top_patient.name, "with", max_count, "diseases.")


# ==========================================
# Testing with static values
# ==========================================

h = HospitalSystem()

print("--- 1. Adding Patients ---")
h.add_new_patient("P1", "Rahul Sharma", 25, "B+", "9876543210")
h.add_new_patient("P2", "Priya Verma", 40, "O+", "9123456789")
h.add_new_patient("P3", "Amit Kumar", 60, "B+", "9988776655")

print("\n--- 10. Adding Multiple Contacts ---")
p1 = h.search_patient("P1")
if p1 != None:
    p1.add_phone("1111111111")
    p1.add_phone("2222222222")

p2 = h.search_patient("P2")
if p2 != None:
    p2.add_phone("3333333333")

print("\n--- 4. Adding Diseases ---")
if p1 != None:
    p1.add_disease("Fever")
    p1.add_disease("Flu")

p3 = h.search_patient("P3")
if p3 != None:
    p3.add_disease("Fever")
    p3.add_disease("Diabetes")
    p3.add_disease("BP")

print("\n--- 11. Display All Patients ---")
h.display_all_patients()

print("\n--- 5. Removing Disease ---")
if p1 != None:
    p1.remove_disease("Flu")

print("\n--- 6. Display All Recorded Diseases ---")
h.display_all_diseases()

print("\n--- 7. Find Patients with Disease 'Fever' ---")
h.find_patients_by_disease("Fever")

print("\n--- 8. Find Patients with Blood Group 'B+' ---")
h.find_patients_by_blood_group("B+")

print("\n--- 13. Find Patient with Highest Diseases ---")
h.patient_with_most_diseases()

print("\n--- 12. Delete Patient P2 ---")
h.delete_patient("P2")

print("\n--- Display All Patients After Deletion ---")
h.display_all_patients()
```

###TASK 6: Movie Ticket Booking System

```python
# Simple classes to store data
class Movie:

    def __init__(self, movie_name):
        self.movie_name = movie_name


class Customer:

    def __init__(self, c_name, c_phone):
        self.c_name = c_name
        self.c_phone = c_phone


class Booking:

    def __init__(self, b_id, customer_obj, movie_obj, timing, num_seats):
        self.b_id = b_id
        self.customer_obj = customer_obj
        self.movie_obj = movie_obj
        self.timing = timing
        self.num_seats = num_seats


class MovieTicketSystem:

    def __init__(self):
        # 1. Dictionary -> stores movie name as key and list of timings as value
        self.movies_dict = {}

        # 2. List -> stores all booking objects
        self.booking_list = []

        # 3. Set -> stores unique customer names
        self.customer_set = set()

        self.current_id = 1

    # 1. Add a movie
    def add_movie(self, m_name):
        if m_name in self.movies_dict:
            print("Movie is already added!")
        else:
            self.movies_dict[m_name] = []
            print("Added movie:", m_name)

    # 2. Add timings for a movie
    def add_timing(self, m_name, t_time):
        if m_name in self.movies_dict:
            self.movies_dict[m_name].append(t_time)
            print("Added timing", t_time, "for", m_name)
        else:
            print("Movie does not exist!")

    # 3. Add a customer booking
    def add_booking(self, c_name, c_phone, m_name, t_time, seats):
        if m_name not in self.movies_dict:
            print("Movie not found!")
            return

        if t_time not in self.movies_dict[m_name]:
            print("Timing not available!")
            return

        # Make objects
        c = Customer(c_name, c_phone)
        m = Movie(m_name)

        # Add customer to set (set handles uniqueness automatically)
        self.customer_set.add(c_name)

        # Create booking object and add to list
        b = Booking(self.current_id, c, m, t_time, seats)
        self.booking_list.append(b)

        print("Booking successful! ID is:", self.current_id)
        self.current_id = self.current_id + 1

    # 4. Search bookings using the Booking ID
    def search_booking_by_id(self, search_id):
        flag = 0
        for b in self.booking_list:
            if b.b_id == search_id:
                print("--- Booking Found ---")
                print("Booking ID:", b.b_id)
                print("Customer Name:", b.customer_obj.c_name)
                print("Movie:", b.movie_obj.movie_name)
                print("Timing:", b.timing)
                print("Seats:", b.num_seats)
                flag = 1
                break

        if flag == 0:
            print("Booking ID not found!")

    # 5. Search all bookings made by a customer
    def search_bookings_by_customer(self, cust_name):
        count = 0
        print("Bookings for customer:", cust_name)
        for b in self.booking_list:
            if b.customer_obj.c_name == cust_name:
                print(
                    "ID:",
                    b.b_id,
                    "| Movie:",
                    b.movie_obj.movie_name,
                    "| Time:",
                    b.timing,
                    "| Seats:",
                    b.num_seats,
                )
                count = count + 1

        if count == 0:
            print("No bookings found for this customer.")

    # 6. Update the number of tickets
    def update_seats(self, b_id, new_seats_count):
        found = False
        for b in self.booking_list:
            if b.b_id == b_id:
                b.num_seats = new_seats_count
                print("Seats updated to", new_seats_count, "for ID:", b_id)
                found = True
                break

        if found == False:
            print("ID not found to update!")

    # 7. Cancel a booking
    def cancel_booking(self, b_id):
        for b in self.booking_list:
            if b.b_id == b_id:
                self.booking_list.remove(b)
                print("Cancelled booking ID:", b_id)
                return
        print("Cannot cancel, ID not found.")

    # 8. Display all unique movies
    def show_all_movies(self):
        print("--- All Unique Movies ---")
        for m in self.movies_dict:
            print(m)

    # 9. Display all unique customers
    def show_all_customers(self):
        print("--- All Unique Customers ---")
        for c in self.customer_set:
            print(c)

    # 10. Find common movie timings between two movies
    def show_common_timings(self, m1, m2):
        if m1 in self.movies_dict and m2 in self.movies_dict:
            list1 = self.movies_dict[m1]
            list2 = self.movies_dict[m2]

            # convert lists to sets to find common elements
            set1 = set(list1)
            set2 = set(list2)
            common_set = set1.intersection(set2)

            print("Common timings between", m1, "and", m2, ":")
            for time in common_set:
                print(time)
        else:
            print("Movies not found in system.")

    # 11. Display the movie with the highest number of bookings
    def highest_booked_movie(self):
        if len(self.booking_list) == 0:
            print("No bookings made yet.")
            return

        # Frequency count dictionary
        count_dict = {}
        for b in self.booking_list:
            mname = b.movie_obj.movie_name
            if mname in count_dict:
                count_dict[mname] = count_dict[mname] + 1
            else:
                count_dict[mname] = 1

        max_movie = ""
        max_val = 0
        for k in count_dict:
            if count_dict[k] > max_val:
                max_val = count_dict[k]
                max_movie = k

        print("Highest booked movie is:", max_movie, "with", max_val, "bookings")

    # 12. Calculate total tickets sold
    def calculate_total_tickets(self):
        total = 0
        for b in self.booking_list:
            total = total + b.num_seats
        print("Total tickets sold overall:", total)

    # 13. Display all bookings in a readable format
    def print_all_bookings(self):
        print("=== PRINTING ALL BOOKINGS ===")
        if len(self.booking_list) == 0:
            print("No bookings to show.")
        else:
            for b in self.booking_list:
                print(
                    "ID: "
                    + str(b.b_id)
                    + " | Name: "
                    + b.customer_obj.c_name
                    + " | Movie: "
                    + b.movie_obj.movie_name
                    + " | Time: "
                    + b.timing
                    + " | Seats: "
                    + str(b.num_seats)
                )


# --- Driver code to test functions ---
sys = MovieTicketSystem()

# 1 and 2: Add movies and timings
sys.add_movie("Batman")
sys.add_timing("Batman", "10:00 AM")
sys.add_timing("Batman", "02:00 PM")

sys.add_movie("Superman")
sys.add_timing("Superman", "02:00 PM")
sys.add_timing("Superman", "06:00 PM")

# 3: Add bookings
sys.add_booking("John", "1234567890", "Batman", "10:00 AM", 2)
sys.add_booking("Alice", "9876543210", "Superman", "02:00 PM", 3)
sys.add_booking("John", "1234567890", "Superman", "06:00 PM", 1)

# 13: Print all
sys.print_all_bookings()

# 4 and 5: Searching
sys.search_booking_by_id(1)
sys.search_bookings_by_customer("John")

# 6: Update seats
sys.update_seats(1, 5)

# 8 and 9: Unique lists
sys.show_all_movies()
sys.show_all_customers()

# 10: Common timings
sys.show_common_timings("Batman", "Superman")

# 11 and 12: Highest & Total
sys.highest_booked_movie()
sys.calculate_total_tickets()

# 7: Cancel booking
sys.cancel_booking(2)
sys.print_all_bookings()
```
