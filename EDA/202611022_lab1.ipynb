### 1. Write a function that takes a string and returns its reverse without using slicing ([::-1]).
Example:
Input: hello
Output: olleh

```python
def reversestring(temp):
  reversed=""
  for char in temp:
    reversed= char+reversed
  return reversed
reversestring("hello world")
```

### 2. Write a function count_even(lst) that takes a list of integers and returns the number of
even numbers.
Example:
Input: [1, 2, 3, 4, 6]
Output: 3

```python
def count_even(lst):
  count=0
  for i in lst:
    if i%2==0:
      count+=1
  return count

print("enter the list of number seprated by space ")
nums=list(map(int,input().split()))
print(count_even(nums))
```

### 5. Write a function that removes duplicate elements from a list while preserving the original
order.
Example
Input:
[1,2,2,3,4,4,5]
Output:
[1,2,3,4,5]

```python
def remove_duplicate(lst):
  newlist=[]
  for i in lst:
    if i not in newlist:
      newlist.append(i)

  return newlist
print("enter the list of number seprated by space ")
nums=list(map(int,input().split()))
print(remove_duplicate(nums))

```

### 6. Write a program that prints all prime numbers from 1 to N.
Example
Input:
20
Output:
2 3 5 7 11 13 17 19

```python
def prime_checker(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
def primeprinter(num):
  for i in range(0,num+1):
    if prime_checker(i):
      print(i,end=" ")
num=int(input("enter the number till which you want to find prime numbers:"))
primeprinter(num)
```

### 8. Write a function that takes two matrices (lists of lists) of the same size and returns their
sum.
Example
A = [[1,2],[3,4]]
B = [[5,6],[7,8]]
Output:
[[6,8],[10,12]]

```python
def matsum(mat1,mat2):
  sum=[]
  for i in range(len(mat1)):
    row=[]
    for j in range(len(mat1[i])):
      row.append(mat1[i][j]+mat2[i][j])
    sum.append(row)
  return sum

A = [[1,2],[3,4]]
B = [[5,6],[7,8]]
print(matsum(A,B))
```

### 11. Write a function to check whether the given username is valid or not. A username is considered
valid if and only if it satisfies all of the following conditions:\
. Length of the username should be between 6 to 12 characters.\
. It should start with an alphabet (either uppercase or lowercase).\
. It should contain only alphanumeric characters (no special characters allowed).\
. It must contain at least one digit [0-9].\
. It should not contain consecutive digits (e.g., "ablcd" is valid, but "ab123cd" is invalid).

```python
while True:
    password = input("enter your password: ")

    # Rule 1: Length check (6 to 12 characters)
    islen = 6 <= len(password) <= 12

    # Rule 2: Must start with an alphabet letter
    alphastart = password[0].isalpha() if password else False

    # Rule 3: Must be alphanumeric (no special characters/spaces)
    alnum = password.isalnum()

    # Rule 4: Must contain at least one digit
    has_digit = any(char.isdigit() for char in password)

    # Rule 5: Check for consecutive increasing digits (e.g., '12', '45')
    consecutive = False
    for i in range(len(password) - 1):
        if password[i].isdigit() and password[i+1].isdigit():
            if int(password[i+1]) == int(password[i]) + 1:
                consecutive = True
                break  # Exit early since we found a match

    # Evaluate all conditions
    if islen and alphastart and alnum and has_digit and not consecutive:
        print("password accepted")
        break
    else:
        print("not follow rules")
```
