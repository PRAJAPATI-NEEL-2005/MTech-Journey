##Q2. Matplotlib -- Histograms, Box Plots & Pie Charts
Perform the following tasks:

"Distribution of Scores".

• Add vertical lines on the histogram to mark the mean and median values in different colors. Include a legend identifying each line.

• Create a box plot comparing the three groups in box_data. Label each group appropriately and display outliers as red diamond markers.

• Create a pie chart showing the proportions in sizes. Label each slice A through E, explode the first slice outward, and display percentage values on each slice.

• Create a stacked bar chart with two data series: [20, 35, 30, 35, 27] and [25, 32, 34, 20, 25].
Use appropriate colors and add a legend

```python
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=1000)
box_data = [np.random.normal(50, 10, 100),
            np.random.normal(60, 15, 100),
            np.random.normal(45, 8, 100)]
labels = ['Group A', 'Group B', 'Group C']
sizes = [30, 25, 20, 15, 10]
explode = (0.1, 0, 0, 0, 0)
mean = np.mean(data)
median = np.median(data)

plt.figure(figsize=(8, 5))

plt.hist(
    data,
    bins=30,
    color='skyblue',
    edgecolor='black'
)

plt.axvline(
    mean,
    color='red',
    linestyle='--',
    linewidth=2,
    label=f'Mean = {mean:.2f}'
)

plt.axvline(
    median,
    color='green',
    linestyle='-',
    linewidth=2,
    label=f'Median = {median:.2f}'
)

plt.title("Distribution of Scores")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(
    box_data,
    tick_labels=labels,
    flierprops={
        'marker': 'D',
        'markerfacecolor': 'red',
        'markeredgecolor': 'red'
    }
)

plt.title("Comparison of Three Groups")
plt.xlabel("Groups")
plt.ylabel("Values")

plt.show()
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    sizes,
    labels=['A', 'B', 'C', 'D', 'E'],
    explode=explode,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Proportions")

plt.show()
```

```python
data1 = [20, 35, 30, 35, 27]
data2 = [25, 32, 34, 20, 25]

categories = ['A', 'B', 'C', 'D', 'E']

plt.figure(figsize=(8, 5))

plt.bar(
    categories,
    data1,
    color='skyblue',
    label='Series 1'
)

plt.bar(
    categories,
    data2,
    bottom=data1,
    color='orange',
    label='Series 2'
)

plt.title("Stacked Bar Chart")
plt.xlabel("Category")
plt.ylabel("Value")
plt.legend()

plt.show()
```

A box plot clearly reveals summary statistics such as the median, interquartile range (IQR), and specific outliers, making it easy to compare central tendencies across multiple groups. In contrast, a histogram shows the detailed shape of a dataset's distribution, including skewness, modality (e.g., bimodal features), and gaps in data that a box plot masks.

##Q3. Matplotlib -- Subplots, Annotations & Customization
Perform the following tasks:

• Create a figure with three subplots arranged vertically (3 rows, 1 column) with a shared x-axis. Set the figure size to 8 inches by 10 inches.

• In the first subplot, plot the sine curve with a blue dashed line. Add an annotation with an arrow pointing to the peak value, labeling it "Peak".

• In the second subplot, plot the cosine curve with a green dotted line. Set the y-axis limits from -1.5 to 1.5.

• In the third subplot, plot the tangent curve with an orange solid line. Add a dashed grid with 70% opacity.

• Apply tight layout to prevent overlapping elements. Add a main title "Trigonometric Functions" at the top of the figure.

• Create a dual-axis plot showing temperature over time. Plot Celsius values on the left y-axis and Fahrenheit values on the right y-axis. Label both axes clearly.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y3 = np.clip(y3, -10, 10)
time = np.arange(0, 10, 0.1)
temp = 25 + 10 * np.sin(time) + np.random.normal(0, 2, len(time))

# 1-5. Figure with 3 vertical subplots and shared x-axis
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(8, 10), sharex=True)

# Subplot 1: Sine curve with blue dashed line and Peak annotation
ax1.plot(x, y1, color='blue', linestyle='--', label='Sine')
peak_idx = np.argmax(y1)
ax1.annotate('Peak', xy=(x[peak_idx], y1[peak_idx]),
             xytext=(x[peak_idx] + 0.8, y1[peak_idx] - 0.3),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6))
ax1.set_title('Sine Curve')
ax1.set_ylabel('sin(x)')
ax1.legend()
ax1.grid(True)

# Subplot 2: Cosine curve with green dotted line and y-limits [-1.5, 1.5]
ax2.plot(x, y2, color='green', linestyle=':', label='Cosine')
ax2.set_ylim(-1.5, 1.5)
ax2.set_title('Cosine Curve')
ax2.set_ylabel('cos(x)')
ax2.legend()
ax2.grid(True)

# Subplot 3: Tangent curve with orange solid line and 70% opacity dashed grid
ax3.plot(x, y3, color='orange', linestyle='-', label='Tangent')
ax3.set_title('Tangent Curve')
ax3.set_xlabel('X')
ax3.set_ylabel('tan(x)')
ax3.legend()
ax3.grid(True, linestyle='--', alpha=0.7)

fig.suptitle('Trigonometric Functions', fontsize=14)
fig.tight_layout()
plt.show()

# 6. Dual-axis plot: Temperature over time (Celsius vs Fahrenheit)
fig, ax_c = plt.subplots(figsize=(8, 5))
temp_f = temp * 1.8 + 32

ax_c.plot(time, temp, color='blue', label='Celsius')
ax_c.set_xlabel('Time')
ax_c.set_ylabel('Temperature (°C)', color='blue')
ax_c.tick_params(axis='y', labelcolor='blue')

ax_f = ax_c.twinx()
ax_f.plot(time, temp_f, color='red', linestyle='--', label='Fahrenheit')
ax_f.set_ylabel('Temperature (°F)', color='red')
ax_f.tick_params(axis='y', labelcolor='red')

plt.title('Temperature over Time (Dual-Axis Plot)')
ax_c.grid(True)
plt.show()
```

Subplots are essential in EDA because they allow side-by-side visualization of multiple signals or variables on aligned axes without cluttering a single axes frame. A secondary y-axis is useful when comparing two continuous variables measured on completely different scales or units over a shared domain, such as Celsius versus Fahrenheit temperature over time.

##Q4. Seaborn -- Distribution Plots (Univariate Analysis)
Perform the following tasks:

• Set the Seaborn visual style to whitegrid and the color palette to muted.

• Create a histogram of the Age column with 20 bins, overlaid with a KDE curve. Use steelblue color.

• Create a KDE plot for the Salary column. Fill the area under the curve with 50% transparency.

• Create a box plot for the Score column. Add a title and ensure all labels are readable.

• Create a violin plot for the Salary column. Display the inner quartile lines inside the violin shape.

• Create an ECDF plot for the Age column. Add a second ECDF line for the Salary column on the same plot (normalize Salary to 0-100 range first, or create separate subplots).

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

np.random.seed(42)
df = pd.DataFrame({
    'Age': np.random.randint(18, 60, 200),
    'Salary': np.random.normal(50000, 15000, 200),
    'Score': np.random.beta(2, 5, 200) * 100,
    'Department': np.random.choice(['IT', 'HR', 'Sales', 'Finance'], 200)
})

# Set visual style and palette
sns.set_theme(style='whitegrid', palette='muted')

# 1. Histogram of Age with 20 bins overlaid with KDE (steelblue)
plt.figure(figsize=(8, 5))
sns.histplot(df['Age'], bins=20, kde=True, color='steelblue')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

# 2. KDE plot for Salary with 50% fill transparency
plt.figure(figsize=(8, 5))
sns.kdeplot(df['Salary'], fill=True, alpha=0.5, color='teal')
plt.title('KDE Plot of Salary')
plt.xlabel('Salary')
plt.ylabel('Density')
plt.show()

# 3. Box plot for Score
plt.figure(figsize=(8, 5))
sns.boxplot(x=df['Score'], color='coral')
plt.title('Box Plot of Score')
plt.xlabel('Score')
plt.show()

# 4. Violin plot for Salary displaying inner quartile lines
plt.figure(figsize=(8, 5))
sns.violinplot(y=df['Salary'], inner='quartile', color='mediumpurple')
plt.title('Violin Plot of Salary')
plt.ylabel('Salary')
plt.show()

# 5. ECDF plot for Age and normalized Salary
salary_norm = (df['Salary'] - df['Salary'].min()) / (df['Salary'].max() - df['Salary'].min()) * 100

plt.figure(figsize=(8, 5))
sns.ecdfplot(df['Age'], label='Age')
sns.ecdfplot(salary_norm, label='Salary (Normalized 0-100)')
plt.title('ECDF Plot of Age and Normalized Salary')
plt.xlabel('Value')
plt.ylabel('Cumulative Probability')
plt.legend()
plt.show()
```

A violin plot is preferred over a box plot when analyzing complex, multimodal data distributions because it displays the full kernel density shape alongside summary statistics, revealing hidden clusters or multiple peaks. An ECDF plot provides unique insights by showing the exact proportion of observations less than or equal to any value without data binning or loss of resolution.

Perform the following tasks:

• Create a box plot of Salary by Department, grouped by Gender. Use a pastel color palette.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

np.random.seed(42)
df = pd.DataFrame({
    'Department': np.random.choice(['IT', 'HR', 'Sales', 'Finance'], 150),
    'Gender': np.random.choice(['Male', 'Female'], 150),
    'Salary': np.random.normal(55000, 12000, 150),
    'Experience': np.random.randint(1, 15, 150)
})

# 1. Count plot of employees in each Department
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Department', palette='Set2')
plt.title('Employee Count by Department')
plt.xlabel('Department')
plt.ylabel('Count')
plt.show()

# 2. Bar plot of average Salary by Department with SD error bars and Gender hue
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Department', y='Salary', hue='Gender', errorbar='sd', palette='Accent')
plt.title('Average Salary by Department and Gender (with Standard Deviation Error Bars)')
plt.xlabel('Department')
plt.ylabel('Salary')
plt.show()

# 3. Strip plot overlaid on top of box plot
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Department', y='Salary', color='lightgray', showfliers=False)
sns.stripplot(data=df, x='Department', y='Salary', jitter=True, alpha=0.6, color='black')
plt.title('Salary by Department (Strip Plot Overlaid on Box Plot)')
plt.xlabel('Department')
plt.ylabel('Salary')
plt.show()

# 4. Swarm plot of Salary by Department using Set2 palette
plt.figure(figsize=(8, 5))
sns.swarmplot(data=df, x='Department', y='Salary', palette='Set2')
plt.title('Swarm Plot of Salary by Department')
plt.xlabel('Department')
plt.ylabel('Salary')
plt.show()

# 5. Box plot of Salary by Department grouped by Gender (pastel palette)
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Department', y='Salary', hue='Gender', palette='pastel')
plt.title('Salary by Department Grouped by Gender')
plt.xlabel('Department')
plt.ylabel('Salary')
plt.show()

# 6. Violin plot of Experience by Department split by Gender with median white dot
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='Department', y='Experience', hue='Gender', split=True, inner='quartile', palette='muted')
plt.title('Experience by Department Split by Gender')
plt.xlabel('Department')
plt.ylabel('Experience (Years)')
plt.show()
```

A strip plot plots every data point along a categorical axis with optional random jitter, which can cause points to overlap when data density is high. A swarm plot uses an optimization algorithm to position points side-by-side without overlap, revealing the density profile. Strip plots scale better for larger datasets, while swarm plots are ideal for smaller datasets where observing individual points without collision is important.

##Q6. Seaborn -- Relationship & Regression Plots (Bivariate)
Perform the following tasks:

• Create a scatter plot of Study_Hours versus Marks. Color the points by Subject and vary the point size by Attendance.

• Create a line plot showing the relationship between Attendance and Marks. Sort the data by Attendance before plotting and include error shading.

• Create a regression plot of Study_Hours versus Marks with a 95% confidence interval band. Reduce the opacity of the scatter points.

• Create a residual plot for the Study_Hours versus Marks relationship. Add a horizontal reference line at zero in red.

• Create a joint plot of Study_Hours versus Marks with a regression line. Set the plot size to 8 inches.

• Create a pair plot of all numeric columns, colored by Subject. Use KDE plots on the diagonal.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

np.random.seed(42)
df = pd.DataFrame({
    'Study_Hours': np.random.uniform(1, 10, 100),
    'Marks': np.random.uniform(40, 100, 100),
    'Attendance': np.random.uniform(60, 100, 100),
    'Subject': np.random.choice(['Math', 'Science', 'English'], 100)
})
df['Marks'] = df['Study_Hours'] * 5 + df['Attendance'] * 0.3 + np.random.normal(0, 5, 100)

# 1. Scatter plot Study_Hours vs Marks colored by Subject, size by Attendance
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Study_Hours', y='Marks', hue='Subject', size='Attendance', sizes=(20, 200), alpha=0.8)
plt.title('Study Hours vs Marks by Subject and Attendance')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.show()

# 2. Line plot Attendance vs Marks (sorted by Attendance) with error shading
df_sorted = df.sort_values('Attendance')
plt.figure(figsize=(8, 5))
sns.lineplot(data=df_sorted, x='Attendance', y='Marks', color='teal')
plt.title('Relationship between Attendance and Marks')
plt.xlabel('Attendance')
plt.ylabel('Marks')
plt.show()

# 3. Regression plot Study_Hours vs Marks with 95% CI band
plt.figure(figsize=(8, 5))
sns.regplot(data=df, x='Study_Hours', y='Marks', ci=95, scatter_kws={'alpha': 0.4}, line_kws={'color': 'red'})
plt.title('Regression Plot of Study Hours vs Marks (95% CI)')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.show()

# 4. Residual plot for Study_Hours vs Marks with red zero reference line
plt.figure(figsize=(8, 5))
sns.residplot(data=df, x='Study_Hours', y='Marks', color='purple')
plt.axhline(0, color='red', linestyle='--')
plt.title('Residual Plot of Study Hours vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Residuals')
plt.show()

# 5. Joint plot of Study_Hours vs Marks with regression line (size 8 inches)
g_joint = sns.jointplot(data=df, x='Study_Hours', y='Marks', kind='reg', height=8)
g_joint.fig.suptitle('Joint Plot of Study Hours vs Marks', y=1.02)
plt.show()

# 6. Pair plot of numeric columns colored by Subject with KDE on diagonal
g_pair = sns.pairplot(df, hue='Subject', diag_kind='kde')
g_pair.fig.suptitle('Pair Plot of Student Performance Data', y=1.02)
plt.show()
```

A pair plot accelerates initial EDA by showing grid matrix visual summaries of all numeric variable pairs simultaneously. Diagonal plots display univariate probability densities (KDE) to evaluate distribution shape and modality, while off-diagonal scatter plots expose linear/non-linear correlations, clusters, and variable interactions.

##Q7. Seaborn -- Heatmaps, Clustermaps & Correlation
Perform the following tasks:

• Compute and print the correlation matrix of all numeric columns in the dataframe.

• Create a heatmap of the correlation matrix. Display the correlation values inside each cell formatted to 2 decimal places. Add grid lines between cells.

• Use a diverging colormap centered at zero with values ranging from -1 to 1.

• Create a clustermap of the correlation matrix. Set the figure size to 8 by 8 inches and use an appropriate diverging colormap.

• Create a heatmap showing only the lower triangle of the correlation matrix. Mask the upper triangle so it appears blank.

• Create a FacetGrid showing the distribution of column A, split into separate subplots based on quartile ranges of column D.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

np.random.seed(42)
df = pd.DataFrame({
    'A': np.random.normal(50, 10, 100),
    'B': np.random.normal(60, 15, 100),
    'C': np.random.normal(40, 8, 100),
    'D': np.random.normal(70, 12, 100),
    'E': np.random.normal(55, 20, 100)
})
df['B'] = df['A'] * 0.8 + np.random.normal(0, 5, 100)
df['C'] = df['A'] * (-0.6) + np.random.normal(0, 5, 100)
df['D'] = df['B'] * 0.5 + np.random.normal(0, 5, 100)

# 1. Compute and print correlation matrix
corr = df.corr()
print("Correlation Matrix:")
print(corr)

# 2 & 3. Heatmap of correlation matrix (annot 2 decimal places, grid lines, diverging colormap -1 to 1)
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.2f', linewidths=0.5, cmap='coolwarm', vmin=-1, vmax=1, center=0)
plt.title('Correlation Matrix Heatmap')
plt.show()

# 4. Clustermap of correlation matrix (8x8 inches)
g_cluster = sns.clustermap(corr, figsize=(8, 8), cmap='coolwarm', vmin=-1, vmax=1, annot=True, fmt='.2f')
g_cluster.fig.suptitle('Correlation Clustermap', y=1.02)
plt.show()

# 5. Lower triangle heatmap
mask = np.triu(np.ones_like(corr, dtype=bool))
plt.figure(figsize=(8, 6))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', linewidths=0.5, cmap='coolwarm', vmin=-1, vmax=1, center=0)
plt.title('Lower Triangle Correlation Heatmap')
plt.show()

# 6. FacetGrid showing distribution of column A split by quartile ranges of column D
df['D_Quartile'] = pd.qcut(df['D'], q=4, labels=['Q1', 'Q2', 'Q3', 'Q4'])
g_facet = sns.FacetGrid(df, col='D_Quartile')
g_facet.map(sns.histplot, 'A', kde=True, color='indigo')
g_facet.fig.suptitle('Distribution of Column A by D Quartiles', y=1.05)
plt.show()
```

Visualizing correlations via a heatmap is more effective than reading a raw numerical table because color gradients immediately highlight strong positive and negative associations across variable pairs. A clustermap is particularly valuable for high-dimensional feature spaces to discover hierarchical groupings and modular relationships between correlated variables.

##Q8. Plotly -- Interactive Basic Charts
Perform the following tasks:

• Create an interactive line chart showing Sales over Month, with a separate line for each Region. Include markers at each data point.

• Create an interactive bar chart showing total Sales by Region. Display the exact sales value on top of each bar.

• Create an interactive scatter plot of Sales versus Profit. Color points by Region, set bubble size by Customers, and show Month information when hovering over points.

• Create an interactive donut chart showing the proportion of total Sales by Region. Set the hole size to 40% of the radius.

• Create an animated scatter plot of Sales versus Profit that transitions through each Month. Fix the x-axis range from 0 to 350 and y-axis range from 0 to 100 for smooth animation.

• Create an interactive area chart showing cumulative Sales by Month for each Region.

```python
import plotly.express as px
import pandas as pd
import numpy as np

np.random.seed(42)
df = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'] * 3,
    'Sales': np.random.randint(100, 300, 18),
    'Profit': np.random.randint(20, 80, 18),
    'Region': ['North', 'South', 'East'] * 6,
    'Customers': np.random.randint(50, 200, 18)
})

# 1. Interactive line chart showing Sales over Month for each Region with markers
fig1 = px.line(df, x='Month', y='Sales', color='Region', markers=True, title='Sales over Month by Region')
fig1.show()

# 2. Interactive bar chart showing total Sales by Region with text on top
sales_by_region = df.groupby('Region', as_index=False)['Sales'].sum()
fig2 = px.bar(sales_by_region, x='Region', y='Sales', text_auto=True, title='Total Sales by Region')
fig2.show()

# 3. Interactive scatter plot of Sales vs Profit (color Region, size Customers, hover Month)
fig3 = px.scatter(df, x='Sales', y='Profit', color='Region', size='Customers', hover_data=['Month'], title='Sales vs Profit Bubble Chart')
fig3.show()

# 4. Interactive donut chart showing proportion of total Sales by Region (hole 40%)
fig4 = px.pie(sales_by_region, values='Sales', names='Region', hole=0.4, title='Proportions of Sales by Region')
fig4.show()

# 5. Animated scatter plot transitioning through Month (x: 0-350, y: 0-100)
fig5 = px.scatter(df, x='Sales', y='Profit', color='Region', size='Customers',
                  animation_frame='Month', range_x=[0, 350], range_y=[0, 100],
                  title='Animated Sales vs Profit over Month')
fig5.show()

# 6. Interactive area chart showing cumulative Sales by Month for each Region
df['Cumulative_Sales'] = df.groupby('Region')['Sales'].cumsum()
fig6 = px.area(df, x='Month', y='Cumulative_Sales', color='Region', title='Cumulative Sales by Month')
fig6.show()
```

Plotly offers significant advantages over static Matplotlib/Seaborn plots during EDA by enabling interactive features like hover tooltips, dynamic filtering, panning, and zooming without editing underlying code. For instance, hovering over a data point on an interactive bubble chart instantly reveals specific metadata like month or customer count, saving time during exploratory analysis.

```python
import plotly.graph_objects as go
import plotly.express as px
import pandas as pd
import numpy as np

np.random.seed(42)
df = pd.DataFrame({
    'Product': ['A', 'B', 'C', 'D', 'E'] * 4,
    'Quarter': ['Q1', 'Q2', 'Q3', 'Q4'] * 5,
    'Revenue': np.random.randint(50, 200, 20),
    'Cost': np.random.randint(30, 150, 20)
})
df['Profit'] = df['Revenue'] - df['Cost']

prod_df = df.groupby('Product', as_index=False)[['Revenue', 'Cost', 'Profit']].sum()

# 1. Grouped bar chart: Revenue and Cost side by side for each Product
fig1 = go.Figure(data=[
    go.Bar(name='Revenue', x=prod_df['Product'], y=prod_df['Revenue'], marker_color='royalblue'),
    go.Bar(name='Cost', x=prod_df['Product'], y=prod_df['Cost'], marker_color='crimson')
])
fig1.update_layout(barmode='group', title='Grouped Bar Chart: Revenue and Cost by Product')
fig1.show()

# 2. Stacked bar chart of Revenue and Cost
fig2 = go.Figure(data=[
    go.Bar(name='Revenue', x=prod_df['Product'], y=prod_df['Revenue'], marker_color='royalblue'),
    go.Bar(name='Cost', x=prod_df['Product'], y=prod_df['Cost'], marker_color='crimson')
])
fig2.update_layout(barmode='stack', title='Stacked Bar Chart: Revenue and Cost by Product')
fig2.show()

# 3. Bubble chart (x=Revenue, y=Profit, size=Cost, color=Product)
fig3 = px.scatter(df, x='Revenue', y='Profit', size='Cost', color='Product', title='Revenue vs Profit Bubble Chart')
fig3.show()

# 4. Interactive dropdown buttons to toggle bar chart between Revenue, Cost, or Profit
fig4 = go.Figure()
fig4.add_trace(go.Bar(x=prod_df['Product'], y=prod_df['Revenue'], name='Revenue', visible=True))
fig4.add_trace(go.Bar(x=prod_df['Product'], y=prod_df['Cost'], name='Cost', visible=False))
fig4.add_trace(go.Bar(x=prod_df['Product'], y=prod_df['Profit'], name='Profit', visible=False))

fig4.update_layout(
    title='Product Financial Metrics (Dropdown Selection)',
    updatemenus=[{
        'buttons': [
            {'label': 'Revenue', 'method': 'update', 'args': [{'visible': [True, False, False]}, {'title': 'Product Revenue'}]},
            {'label': 'Cost', 'method': 'update', 'args': [{'visible': [False, True, False]}, {'title': 'Product Cost'}]},
            {'label': 'Profit', 'method': 'update', 'args': [{'visible': [False, False, True]}, {'title': 'Product Profit'}]}
        ],
        'direction': 'down',
        'showactive': True
    }]
)
fig4.show()

# 5. 3D scatter plot (Revenue on x, Cost on y, Profit on z, color Product)
fig5 = px.scatter_3d(df, x='Revenue', y='Cost', z='Profit', color='Product', title='3D Scatter Plot: Revenue vs Cost vs Profit')
fig5.show()

# 6. Interactive table displaying first 10 rows
fig6 = go.Figure(data=[go.Table(
    header=dict(values=list(df.columns), fill_color='paleturquoise', align='left'),
    cells=dict(values=[df[col].iloc[:10] for col in df.columns], fill_color='lavender', align='left')
)])
fig6.update_layout(title='First 10 Rows of Product Data')
fig6.show()
```

Plotly Graph Objects is preferred over Plotly Express when building custom, multi-layered visualizations that require precise control over UI widgets like dropdown menus, layout parameters, or complex trace configurations. The main trade-off is that Graph Objects requires writing more verbose, lower-level code compared to the concise syntax of Plotly Express.

##Q10. Plotly -- Interactive EDA Dashboards & Advanced Charts
Perform the following tasks:

• Create an interactive sunburst chart showing the hierarchy from City to Category, with Sales as the value. Color the segments by Profit using a color scale.

• Create an interactive treemap with the same City-to-Category hierarchy. Display the average Rating when hovering over each segment.

• Create a faceted scatter plot of Sales versus Profit with a separate subplot for each Quarter. Color points by Category.

• Create a 2x2 multi-plot dashboard containing: (1) a histogram of Sales, (2) a box plot of Profit by Category, (3) a scatter plot of Sales versus Profit, and (4) a bar chart of average Sales by City. Add subplot titles.

• Create an animated bar chart showing Sales by Category that transitions through each Quarter. Fix the y-axis range to accommodate the maximum value across all quarters.

• Create an interactive violin plot of Sales by Category, grouped by City. Display a box plot inside each violin and show all individual data points.

```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np

np.random.seed(42)
n = 300
df = pd.DataFrame({
    'City': np.random.choice(['New York', 'LA', 'Chicago', 'Houston', 'Phoenix'], n),
    'Category': np.random.choice(['Electronics', 'Clothing', 'Food', 'Home'], n),
    'Sales': np.random.lognormal(4, 0.5, n),
    'Profit': np.random.lognormal(3, 0.4, n),
    'Rating': np.random.choice([1, 2, 3, 4, 5], n, p=[0.05, 0.1, 0.2, 0.35, 0.3]),
    'Quarter': np.random.choice(['Q1', 'Q2', 'Q3', 'Q4'], n)
})

# 1. Sunburst chart (City -> Category, value=Sales, color=Profit)
fig1 = px.sunburst(df, path=['City', 'Category'], values='Sales', color='Profit',
                  color_continuous_scale='RdYlGn', title='Sales Hierarchy (Sunburst Chart)')
fig1.show()

# 2. Treemap (City -> Category, hover average Rating)
df_tree = df.groupby(['City', 'Category'], as_index=False).agg({'Sales': 'sum', 'Rating': 'mean', 'Profit': 'sum'})
fig2 = px.treemap(df_tree, path=['City', 'Category'], values='Sales', hover_data=['Rating'],
                  color='Profit', title='Sales & Average Rating Treemap')
fig2.show()

# 3. Faceted scatter plot of Sales vs Profit by Quarter, colored by Category
fig3 = px.scatter(df, x='Sales', y='Profit', facet_col='Quarter', color='Category',
                  title='Sales vs Profit Faceted by Quarter')
fig3.show()

# 4. 2x2 multi-plot dashboard
fig4 = make_subplots(rows=2, cols=2, subplot_titles=('Sales Histogram', 'Profit by Category (Box)',
                                                     'Sales vs Profit Scatter', 'Average Sales by City'))

fig4.add_trace(go.Histogram(x=df['Sales'], name='Sales Hist', marker_color='indigo'), row=1, col=1)

for cat in df['Category'].unique():
    fig4.add_trace(go.Box(y=df[df['Category'] == cat]['Profit'], name=cat), row=1, col=2)

fig4.add_trace(go.Scatter(x=df['Sales'], y=df['Profit'], mode='markers', marker=dict(color=df['Rating'], colorscale='Viridis'), name='Scatter'), row=2, col=1)

city_sales = df.groupby('City', as_index=False)['Sales'].mean()
fig4.add_trace(go.Bar(x=city_sales['City'], y=city_sales['Sales'], name='Avg Sales', marker_color='teal'), row=2, col=2)

fig4.update_layout(height=800, width=900, title_text='Interactive EDA Dashboard', showlegend=False)
fig4.show()

# 5. Animated bar chart showing Sales by Category transitioning through Quarter
cat_qtr = df.groupby(['Category', 'Quarter'], as_index=False)['Sales'].sum()
max_sales = cat_qtr['Sales'].max() * 1.1
fig5 = px.bar(cat_qtr, x='Category', y='Sales', animation_frame='Quarter',
              range_y=[0, max_sales], title='Sales by Category across Quarters (Animated)')
fig5.show()

# 6. Interactive violin plot of Sales by Category grouped by City with box & points
fig6 = px.violin(df, x='Category', y='Sales', color='City', box=True, points='all',
                 title='Sales Distribution by Category and City (Violin Plot)')
fig6.show()
```

Interactive dashboards improve EDA by presenting complementary statistical views (distributions, correlations, and categorical breakdowns) side by side, enabling holistic pattern discovery without constant plot switching. A sunburst or treemap chart is much more effective than a standard bar chart when exploring hierarchical, multi-level datasets where understanding proportion across nested levels is key.
