##Q6. GroupBy Operations

```python
df = pd.DataFrame({
'Department': ['Sales', 'IT', 'Sales', 'HR', 'IT', 'HR', 'Sales', 'IT'],
'Employee': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace', 'Henry'],
'Salary': [50000, 70000, 55000, 45000, 75000, 48000, 60000, 72000],
'Experience': [2, 5, 3, 4, 6, 3, 4, 7]
})
df.to_csv('employee_data.csv', index=False)
df = pd.read_csv('employee_data.csv')
```

```python
mean = df.groupby('Department')['Salary'].mean()
print(mean)
```

```python
totalsalary = df.groupby('Department')['Salary'].sum()
print(totalsalary)
```

```python
employeecount = df.groupby('Department').size()
print(employeecount)
```

```python
exp = df.groupby('Department')['Experience'].agg(['min', 'max', 'mean'])
print(exp)
```

```python
dept = df.groupby('Department').agg({
    'Salary': 'mean',
    'Experience': 'sum'
})
print(dept)
```

```python
df['Dept_Avg_Salary'] = df.groupby('Department')['Salary'].transform('mean')
print(df)
```

```python
deptavg = df.groupby('Department')['Salary'].mean()
filtered_departments = deptavg[deptavg > 55000]
print(filtered_departments)
```

agg() reduces each group to a summary value, such as one mean salary per department. transform() calculates something for each group but returns the result with the same number of rows as the original DataFrame, so it can be added back as a column.

##Q8. Data Transformation: Apply, Map, and Replace

```python
df = pd.DataFrame({
'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
'Marks': [85, 78, 92, 88, 75],
'Grade_Letter': ['B', 'C', 'A', 'A', 'C']
})
df.to_csv('grades_data.csv', index=False)
df = pd.read_csv('grades_data.csv')
```

```python

df['Marks_Squared'] = df['Marks'].apply(lambda x: x**2)
print(df)
```

```python
df['Category'] = df['Marks'].apply(
    lambda x: 'High' if x >= 85
    else 'Medium' if x >= 70
    else 'Low'
)
print(df)


```

```python
grade_map = {
    'A': 4,
    'B': 3,
    'C': 2,
    'D': 1,
    'F': 0
}

df['Grade_Points'] = df['Grade_Letter'].map(grade_map)
print(df)
```

```python
df['Name'] = df['Name'].replace('Charlie', 'Chris')
print(df)
```

```python
df = df.apply(
    lambda column: column.map(
        lambda x: x.upper() if isinstance(x, str) else x
    )
)
print(df)
```

apply() is mainly used on a Series or DataFrame when you want to apply a function to values, rows, or columns. map() is used on a Series to transform individual values, while applymap() was used to apply a function to every individual cell of a DataFrame; in newer Pandas versions, DataFrame.map() is used instead of applymap()

##Q9. Merging, Joining, and Concatenating

```python
df1 = pd.DataFrame({
'ID': [1, 2, 3, 4],
'Name': ['Alice', 'Bob', 'Charlie', 'Diana']
})
df1.to_csv('employees.csv', index=False)
df2 = pd.DataFrame({
'ID': [3, 4, 5, 6],
'Salary': [50000, 60000, 70000, 80000]
})
df2.to_csv('salaries.csv', index=False)
df3 = pd.DataFrame({
'ID': [7, 8],
'Name': ['Eve', 'Frank'],
'Salary': [55000, 65000]
})
df3.to_csv('new_employees.csv', index=False)
df1 = pd.read_csv('employees.csv')
df2 = pd.read_csv('salaries.csv')
df3 = pd.read_csv('new_employees.csv')
```

```python
inner = pd.merge(df1, df2, on='ID', how='inner')
print("Inner Join:")
print(inner)
print("Rows:", len(inner))


```

```python
left = pd.merge(df1, df2, on='ID', how='left')
print("\nLeft Join:")
print(left)
```

```python
right = pd.merge(df1, df2, on='ID', how='right')
print("\nRight Join:")
print(right)
```

```python
outer = pd.merge(df1, df2, on='ID', how='outer')
print("\nOuter Join:")
print(outer)
print("Rows:", len(outer))
```

```python
vertical = pd.concat([df1, df3], axis=0)
print("\nVertical Concatenation:")
print(vertical)
```

```python
horizontal = pd.concat([df1, df2], axis=1)
print("\nHorizontal Concatenation:")
print(horizontal)
```

merge() is used when you want to combine DataFrames based on a common column or key, such as ID. concat() is used to simply join DataFrames together row-wise or column-wise, usually when their structure is compatible.

Use an inner join when you only want records that exist in both datasets and are interested in their common information. Use an outer join during EDA when you want to see all records and identify missing matches, which is useful for finding data-quality or missing-data problems

##Q10. Pivot Tables and Cross Tabulation

```python
df = pd.DataFrame({
'Region': ['North', 'South', 'East', 'West', 'North',
'South', 'East', 'West', 'North', 'South'],
'Product': ['A', 'A', 'B', 'B', 'A', 'B', 'A', 'B', 'B', 'A'],
'Sales': [200, 150, 300, 250, 180, 220, 280, 260, 190, 210],
'Profit': [50, 30, 80, 60, 45, 55, 70, 65, 40, 52]
})
df.to_csv('sales_data.csv', index=False)
df = pd.read_csv('sales_data.csv')
```

```python
pivot_sales = pd.pivot_table(
    df,
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='mean'
)

print(pivot_sales)
```

```python
pivot_profit = pd.pivot_table(
    df,
    values='Profit',
    index='Region',
    columns='Product',
    aggfunc='sum'
)

print(pivot_profit)
```

```python
pivot_multiple = pd.pivot_table(
    df,
    values=['Sales', 'Profit'],
    index='Region',
    aggfunc={
        'Sales': 'mean',
        'Profit': 'sum'
    }
)

print(pivot_multiple)
```

```python
cross_count = pd.crosstab(
    df['Region'],
    df['Product']
)

print(cross_count)
```

```python
cross_proportion = pd.crosstab(
    df['Region'],
    df['Product'],
    normalize=True
)

print(cross_proportion)
```

```python
cross_margin = pd.crosstab(
    df['Region'],
    df['Product'],
    margins=True
)

print(cross_margin)
```

groupby() is useful when you mainly want to calculate summaries for groups. pivot_table() is more useful in EDA when you want to see relationships between two or more categorical variables in a table format, making comparisons easier.

A pivot table is mainly used to summarize numerical data using functions like mean, sum, or count. A crosstab is mainly used to show the frequency or proportion of combinations between categorical variables, such as how many Product A and Product B records exist in each Region.
