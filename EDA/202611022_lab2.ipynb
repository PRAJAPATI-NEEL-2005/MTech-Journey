##Q7. Boolean Masking and Conditional Data Modification  
Consider the following temperature dataset: \
temperature = np.array([ \
32, 28, 35, 40, 25, \
30, 42, 27, 38 \
]) \
Using NumPy Boolean masking: \
• Extract all temperatures greater than 35. \
• Extract temperatures between 28 and 38 inclusive. \
• Count how many temperatures are greater than 35. \
• Count how many temperatures are below 30. \
• Replace every temperature greater than 40 with 40. \
• Replace every temperature below 28 with 28. \
• Display the final modified array.

```python
temperature = np.array([32, 28, 35, 40, 25,30, 42, 27, 38])
tempgreat=temperature[temperature>35]
print("temperatures greater than 35:",tempgreat)
tempbetween=temperature[(temperature>=28) & (temperature<=38)]
print("temperatures between 28 and 38:",tempbetween)
print("num of temperatures greater than 35:",len(tempgreat))
print("num of temperatures less than 30:",len(temperature[temperature<30]))
temperature[temperature>40]=40
temperature[temperature<28]=28
print("modified temperature:",temperature)
```

## Q10. Vectorized Grade Assignment  

```python
marks = np.array([35, 78, 92, 41, 56, 29, 88, 67])

conditions = [
    (marks >= 90) & (marks <= 100),
    (marks >= 80) & (marks <= 89),
    (marks >= 70) & (marks <= 79),
    (marks >= 50) & (marks <= 69),
    (marks >= 35) & (marks <= 49),
    (marks < 35)
]

choices = ['A+', 'A', 'B', 'C', 'D', 'F']

grades = np.select(conditions, choices, default='')

print("Student Grades:")
for mark, grade in zip(marks, grades):
    print(f"Mark: {mark} -> Grade: {grade}")
unique_grades, counts = np.unique(grades, return_counts=True)
grade_counts = dict(zip(unique_grades, counts))

print("\nGrade Category Counts:")
for grade in choices:
    print(f"Grade {grade}: {grade_counts.get(grade, 0)} student(s)")
```
