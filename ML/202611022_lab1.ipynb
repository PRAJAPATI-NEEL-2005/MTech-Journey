2. Write a function to calculate the mean, minimum and maximum of a list.

```python
def mean_min_max(arr):
  if len(arr)==0:
    return None
  minele = arr[0]
  maxele = arr[0]
  sumele = 0
  for ele in arr:
    if ele<minele:
      minele = ele
    if ele>maxele:
      maxele = ele
    sumele += ele
  meanele = sumele/len(arr)
  temp =[meanele,minele,maxele]
  return temp
arr =[51,41,84,95,65,85]
mean_min_max = mean_min_max(arr)
if mean_min_max==None:
  print("list is empty")
else:
  print("mean : ",end ="")
  print(mean_min_max[0])
  print("minimum : ",end = "")
  print(mean_min_max[1])
  print("maximum : ",end = "")
  print(mean_min_max[2])
```

##Dataset
Use the California Housing Dataset available in sklearn.datasets.

```python
from sklearn.datasets import fetch_california_housing
import numpy as np
housing_data = fetch_california_housing()
feature = housing_data.data
target = housing_data.target
```

# Task 3: Dataset Preparation

1. Split the dataset into:
*   Training Set (60%)
*   Validation Set (20%)
*   Test Set (20%)

```python
rows = feature.shape[0]
np.random.seed(1)
shuffle = np.random.permutation(rows)
train_size =int(0.6 * rows)
validation_size = int (0.2 * rows)
train_indexes = shuffle[:train_size]
validation_indexes = shuffle[train_size:train_size+validation_size]
test_indexes = shuffle[train_size+validation_size:]
x_train = feature[train_indexes]
y_train = target[train_indexes]

x_validation = feature[validation_indexes]
y_validation = target[validation_indexes]

x_test = feature[test_indexes]
y_test = target[test_indexes]
```

2. Standardize the input features using only the training-set mean and standard deviation

```python
train_mean = np.mean(x_train,axis=0)
train_std = np.std(x_train,axis=0)
x_train_standardized =  (x_train-train_mean)/train_std
x_val_standardized = (x_validation-train_mean)/train_std
x_test_standardized = (x_test-train_mean)/train_std
print(x_train_standardized)
print(x_val_standardized)
print(x_test_standardized)
```

3. Add a column of ones to represent the bias parameter.

```python
def addbias(x):
  bias = np.ones((x.shape[0],1))
  return np.hstack((bias,x))

x_train_final = addbias(x_train_standardized)
x_val_final = addbias(x_val_standardized)
x_test_final = addbias(x_test_standardized)
print(x_train_final[0])
print(x_val_final[0])
print(x_test_final[0])
```

## Task 4: Implement Linear Regression

Implement Linear Regression from scratch using NumPy and Batch Gradient Descent. In
Batch Gradient Descent, the entire training set is used to compute the gradient before updating the model parameters once per epoch.

```python
def calculate_loss(x,y,theta):
  predict = x @ theta
  errors = predict - y
  mse = (1/len(y))* (np.sum(errors **2))
  return mse
def batch_gradient_descent (x_train,y_train,x_val,y_val,learning,epochs):
  theta = np.zeros(x_train.shape[1])
  train_loss_history =[]
  val_loss_history = []
  for epoch in range (epochs) :
    predict = x_train @ theta
    errors = predict - y_train
    gradient =(2/x_train.shape[0]) * (x_train.T @ errors)

    theta = theta - learning * gradient

    train_mse = calculate_loss(x_train,y_train,theta)
    train_loss_history.append(train_mse)

    val_mse = calculate_loss(x_val,y_val,theta)
    val_loss_history.append(val_mse)

  return train_loss_history,val_loss_history
batch_gradient_descent(x_train_final,y_train,x_val_final,y_validation,0.05,10)
```

Train the implemented model with each of the following learning rates:

Write a conclusion explaining:

• What the minimum point of the curve represents.

The lowest point on both U-shaped curves represents the Global Minimum of the cost function. At this exact point, the model has found the optimal weights. where the MSE is at its absolute lowest.

• Why the loss increases when the parameter moves away from the optimum.

As the parameter shift left or right from the optimum mark point the model's prediction diverge from the actual target value. Because the MSE formula squres the errorso that any small shift away from the correct weight causes the total error penalty to increase exponentially

• How Gradient Descent moves towards the minimum.

Gradient Descent calculates the slope at any random starting point on this curve. If the parameter is on the right side of the curve, the slope is positive, so the algorithm steps left. If it's on the left side, the slope is Negative, so it steps right. Over multiple epochs, it rolls down the slope until it reaches the flat bottom where the gradient is zero.
